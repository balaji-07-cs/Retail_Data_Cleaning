# Retail Store Sales Data Cleaning and Preprocessing

## Project Overview

This project focuses on cleaning and preprocessing a real-world retail store sales dataset using Python and Pandas.

The dataset contains transaction details such as customer information, product categories, prices, quantities, payment methods, locations, transaction dates, and discount information.

The main objective is to identify and handle missing values, check duplicate records, correct data types, detect potential outliers, perform feature engineering, and export a clean dataset for further analysis.

## Project Objective

The objective of this project is to prepare a messy retail dataset for further analysis by:

- Loading the raw dataset using Pandas
- Identifying missing values and handling them appropriately
- Checking and removing duplicate records
- Correcting incorrect data types
- Detecting potential outliers
- Creating useful features from transaction dates
- Calculating revenue from price and quantity
- Exporting the final cleaned dataset as `clean_dataset.csv`

```python
Q1 = df["Total Spent"].quantile(0.25)
Q3 = df["Total Spent"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
outliers = df[df["Total Spent"] > upper_limit]

print("Number of potential outliers:", len(outliers))
```

```python
outliers[["Transaction ID", "Price Per Unit", "Quantity", "Total Spent"]]
```

```python
df.isnull().sum()
```

```python
df["Item"] = df["Item"].fillna(df["Item"].mode()[0])
```

```python
df["Price Per Unit"] = df["Price Per Unit"].fillna(
    df["Price Per Unit"].median()
)
```

```python
df["Quantity"] = df["Quantity"].fillna(
    df["Quantity"].median()
)
```

```python
df["Total Spent"] = df["Total Spent"].fillna(
    df["Price Per Unit"] * df["Quantity"]
)
```

```python
df["Discount Applied"] = df["Discount Applied"].fillna(False)
```

```python
df.isnull().sum()
```

```python
df["Transaction Date"] = pd.to_datetime(
    df["Transaction Date"],
    errors="coerce"
)
```

```python
print(df["Transaction Date"].dtype)
```

```python
print("Invalid dates:", df["Transaction Date"].isnull().sum())
```

```python
df["Year"] = df["Transaction Date"].dt.year
```

```python
df["Month"] = df["Transaction Date"].dt.month
```

```python
df["Month Name"] = df["Transaction Date"].dt.month_name()
```

```python
df["Revenue"] = df["Price Per Unit"] * df["Quantity"]
```
